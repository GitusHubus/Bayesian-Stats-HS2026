```python
import numpy as np;
import matplotlib.pyplot as plt;

#Exercise 1

def p(x):
        return 1/2 *np.sin(x)*np.heaviside(np.pi - x,0)*np.heaviside(x,0);

def P(x):
    return 1/2*(1- np.cos(x))*np.heaviside(x,0)*np.heaviside(np.pi - x,0)+np.heaviside(x-np.pi,1);
    
test = np.arange(-np.pi,2*np.pi,0.001)

plt.plot(test,p(test))
plt.plot(test,P(test))
plt.show()

def Pinv(u):
    return np.arccos(1-2*u);
    
u = np.random.uniform(0,1,10000);
X = Pinv(u);


plt.hist(u,bins=20)
plt.show()

plt.hist(X,bins=20)
plt.show()
```
