```python
#Sum
def my_function(n):
    x=0;
    k=0;
    while k < n:
        x += 4*(-1)**k/(2*k+1);
        k = k+1;
    print(x)
    
    
```

```python
my_function(1000)
```

```python
#Square and such

Points_x = np.random.uniform(-1,1,10000)
Points_y = np.random.uniform(-1,1,10000)
          
i=0;
coordinates = [];

while i<len(Points_x):
    coordinates.append((Points_x[i],Points_y[i]));
    i+=1;

i=0;
n_circ=0

while i<len(coordinates):
    if np.sqrt(coordinates[i][0]**2 + coordinates[i][1]**2) <= 1:
        n_circ = n_circ + 1;
    i = i + 1;
        

print(n_circ)  
print(len(coordinates))
# If r=1, A_circ = Pi and A_square = 4. A_circ/A_square = Pi/4 = n_circ/n_tot
print(4*n_circ/len(coordinates))


inside = (Points_x**2 + Points_y**2)<1;

plt.scatter(Points_x[inside],Points_y[inside],color = "red",s=1)
plt.scatter(Points_x[~inside],Points_y[~inside],color = "blue",s=1)
                
```
