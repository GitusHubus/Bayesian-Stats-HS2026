```python
import numpy as np
import matplotlib.pyplot as plt
import scipy
```

```python
def target(x):
    return np.exp(-x**2);
def proposal(x):
    return np.piecewise(
        x,
        [x < -5, (x >= -5) & (x <= 5), x > 5],     # conditions
        [lambda x: 0, 1.0, lambda x: 0])
def proposal_CDF(x):
    return np.piecewise(
        x,
        [x < -5, (x >= -5) & (x <= 5), x > 5],     # conditions
        [lambda x: 0, x, lambda x: 1])
y = np.linspace(-10,10)
```

```python
plt.plot(y,target(y))
plt.plot(y,proposal(y))
```
