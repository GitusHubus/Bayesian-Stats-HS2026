```python
import h5py
import numpy as np
from pathlib import Path
```

```python
#Check file names

DATA = Path("Data")
files = sorted(DATA.glob("*.h5"))
for f in files:
    print(f.name)
```

```python
#Check structure

def show(name, obj):
    if isinstance(obj, h5py.Dataset):
        print(f"  {name:50s} shape={obj.shape}")

for f in files:
    print(f.name)
    with h5py.File(f) as h:
        h.visititems(show)
    print()
```

```python
spec_file = files[1]
cov_file = files[0]
win_file = files[2]

def k_info(f, prefix):
    with h5py.File(f) as h:
        for ell in ["0", "2", "4"]:
            k = h[f"{prefix}{ell}/k"][:]
            edges = h[f"{prefix}{ell}/k_edges"][:]
            width = np.median(edges[:, 1] - edges[:, 0])
            print(f"  ell={ell}: {len(k)} bins, k = {edges[0,0]:.3f} … {edges[-1,1]:.3f}, bin width = {width:.4f}")

print("spectrum");   k_info(spec_file, "")
print("covariance"); k_info(cov_file, "observable/")
print("window");     k_info(win_file, "observable/")
```
